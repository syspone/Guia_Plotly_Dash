Objetivo: transformar o total de vendas por país em um gráfico de barras com Plotly e exportá-lo como PNG.

## 1. Preparar os dados

```python
from dashvendas.data import load_sales, total_by_country
from dashvendas.visualization import sales_by_country_chart, save_figure

totals = total_by_country(load_sales())
totals
```

## 2. Gráfico de barras horizontais
O mesmo gráfico usado no dashboard (`src/dashvendas/visualization.py`).

```python
fig = sales_by_country_chart(totals)
fig
```

## 3. Exportar como imagem
O PNG vai para `reports/figures/` (exportação feita pelo Kaleido).

```python
save_figure(fig)
```
